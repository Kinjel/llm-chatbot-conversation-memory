```python
from groq import Groq
import os

# API setup
GROQ_API_KEY = "your_key_here"

client = Groq(api_key=GROQ_API_KEY)
```

```python
# Chat class
class ChatNode:

    def prep(self, shared):
        # Add user input
        shared["messages"].append({"role": "user", "content": user_input})
        shared["messages"].append({"role": "assistant", "content": ""})
        return shared["messages"]

    def exec(self, prep_res):
        # Model call
        response = client.chat.completions.create(
            messages=prep_res,
            model="llama-3.3-70b-versatile"
        )
        return response.choices[0].message.content

    def post(self, shared, prep_res, exec_res):
        # Save response
        shared["messages"][-1]["content"] = exec_res

    def run(self, shared):
        prep_res = self.prep(shared)
        exec_res = self.exec(prep_res)
        self.post(shared, prep_res, exec_res)
        print("\nBot:", exec_res)
```

```python
# Initialize memory
shared = {}
shared["messages"] = [
    {
        "role": "system",
        "content": "You are a helpful AI assistant."
    }
]

chat_node = ChatNode()
```

```python
# Start chatbot
while True:
    user_input = input("\nAsk something (type exit to stop): ")

    if user_input.lower() == "exit":
        print("Chat ended.")
        break

    chat_node.run(shared)
```
